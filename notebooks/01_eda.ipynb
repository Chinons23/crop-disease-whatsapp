```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

DATA_ROOT = Path('../data/raw/PlantVillage')
```

```python
# Count images per class
counts = {
    d.name: len(list(d.glob('*')))
    for d in DATA_ROOT.iterdir() if d.is_dir()
}
counts_df = pd.DataFrame(sorted(counts.items()), columns=['class', 'count'])
counts_df
```

```python
counts_df.plot(kind='bar', x='class', y='count', figsize=(14, 5), legend=False)
plt.ylabel('# images')
plt.title('Class distribution')
plt.tight_layout()
plt.show()
```
