```python
import sys
import os


# 1. Walk through /kaggle/input to find where main.py actually is
code_dir = None
for root, dirs, files in os.walk('/kaggle/input'):
    if 'main.py' in files:
        code_dir = root
        break

if code_dir:
    print(f"Found main.py in: {code_dir}")
    # Force Python to look in the directory where main.py lives
    if code_dir not in sys.path:
        sys.path.insert(0, code_dir)
        
    # Optional: Change working directory there if your script uses relative assets
    os.chdir(code_dir)
else:
    print("ERROR: main.py could not be found anywhere in /kaggle/input!")
    print("Available paths:", [os.path.join(r, f) for r, d, fs in os.walk('/kaggle/input') for f in fs])

# 2. Install dependencies
%pip install timm PyYAML

# Pointing Argparse directly to train_config.yaml
sys.argv = ['main.py', '--config', 'configs/train_config.yaml']

import os, glob, yaml
base = "/kaggle/input/datasets/chinonsosilas/crop-disease-v2"
for f in glob.glob(f"{base}/configs/*.y*ml"):
    print(f, "->", yaml.safe_load(open(f)).get("data_root"))

print(glob.glob("/kaggle/input/**/color", recursive=True))



color = glob.glob("/kaggle/input/**/color", recursive=True)[0]
src_cfg = f"{base}/configs/train_config.yaml"
cfg = yaml.safe_load(open(src_cfg))
cfg["data_root"] = color
yaml.safe_dump(cfg, open("/kaggle/working/train_config.yaml", "w"))

sys.path.insert(0, base)
import main
main.main(["--config", "/kaggle/working/train_config.yaml"])
```
